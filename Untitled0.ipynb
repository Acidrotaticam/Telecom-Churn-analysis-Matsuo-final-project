```python
telecom2 = telecom.dropna(subset=["change_mou", "rev_Mean"])
telecom2 = telecom2.fillna(telecom2.mean(numeric_only=True))

telecom2 = telecom2.fillna('Unknown')

telecom2['total_drop'] = (
    telecom2['drop_vce_Mean'] +
    telecom2['drop_dat_Mean']
)
telecom2["total_blck"] = (
    telecom2['blck_vce_Mean'] +
    telecom2['blck_dat_Mean']
)
telecom2['drop_block_ratio'] = telecom2['drop_blk_Mean'] / (telecom2['attempt_Mean'] + 0.01)
telecom2['avg_custcare_duration'] = telecom2['cc_mou_Mean'] / (telecom2['custcare_Mean'] + 0.01)

telecom2['min_trend'] = telecom2['avg3mou'] / (telecom2['avg6mou'] + 0.01)
telecom2['rev_trend'] = telecom2['avg3rev'] / (telecom2['avg6rev'] + 0.01)


null = telecom2.isnull().sum()
null = null[null > 0]
print(f"number of nulls: {null}")
```

```python
high_importance = ['drop_vce_Mean', 'drop_dat_Mean', 'blck_vce_Mean', 'blck_dat_Mean' ]

for col in high_importance:
    telecom2[f'{col}_log'] = np.log1p(telecom2[col])

telecom2['churn_label'] = telecom2['churn'].map({0: 'stayed', 1: 'churned'})

high_importance = ["eqpdays", 'drop_vce_Mean_log', 'drop_dat_Mean_log', 'blck_vce_Mean_log', 'blck_dat_Mean_log' ]

plot_cols = high_importance + ['churn_label']
existing_cols = [c for c in plot_cols if c in telecom2.columns]
pair_sample = telecom2[existing_cols].sample(n=5000, random_state=42)

sb.pairplot(pair_sample, hue='churn_label', corner=True, diag_kind='kde')
pt.show()
```

```python
for col in telecom2.select_dtypes(include=['object']).columns:
    telecom2[col] = telecom2[col].astype('category')

X = telecom2.drop(['churn',"Customer_ID", "churn_label"], axis=1)
y = telecom2['churn']

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
model = XGBClassifier(
    n_estimators=380,
    learning_rate=0.03,
    max_depth=7,
    eval_metric='auc',
    random_state=42,
    enable_categorical=True
)

auc_scores = cross_val_score(model, X, y, cv=cv, scoring='roc_auc')

print(f"Average Validation AUC: {round(np.mean(auc_scores), 4)}")
print(f"Standard Deviation: {round(np.std(auc_scores), 4)}")

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

cm = confusion_matrix(y_test, y_pred)

pt.figure(figsize=(5, 4))
sb.heatmap(cm, annot=True, fmt='d', xticklabels=['stay', 'churn'], yticklabels=['stay', 'churn'])
pt.title('Confusion matrix')
pt.xlabel('Predicted')
pt.ylabel('Actual')
pt.show()
```

## Analysis

$H_0: \mu_{\text{stayed}} = \mu_{\text{churned}}$
$H_1: \mu_{\text{stayed}} \neq \mu_{\text{churned}}$

* **Statistical method:** Two-sample Welch's t-test (Unpaired/Independent)
* **Significance level:** $\alpha = 0.05$
* **Effect size check:** Cohen's $d$

---

### Justification:
The data meets the minimum requirements for both the T-test and Cohen's d:
1. **Independence of samples**
2. **Churn is a binary data type**
3. **Sufficient amount of data for normal approximation**

```python
df_analisis = telecom2.copy()

print("== Variables ==")
print(df_analisis[['churn'] + columns].head())

def cohen_d(group1, group2):
    diff = group1.mean() - group2.mean()
    n1, n2 = len(group1), len(group2)
    var1, var2 = group1.var(ddof=1), group2.var(ddof=1)
    pooled_std = np.sqrt(((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2))
    return abs(diff / pooled_std)

for col in columns:
    print(col)

    df_temp = telecom2[telecom2[col] < limits[col]]

    grouped = df_temp.groupby('churn')[col]
    stayed, churned = [group for _, group in grouped]

    print(f"Stayed mean: {stayed.mean():.2f} | Churned mean: {churned.mean():.2f} | Residual: {churned.mean() - stayed.mean():.2f}")

    _, t_pval = stats.ttest_ind(stayed, churned, equal_var=False)
    _, u_pval = stats.mannwhitneyu(stayed, churned, alternative='two-sided')

    d_val = cohen_d(stayed, churned)
    print(f"-> [Cohen's D]: {d_val:.3f}")

    print(f"-> [T-test] p-value: {t_pval:.4e} → {'Reject H_0' if t_pval < 0.05 else 'H_0 not rejected'}")
    print(f"-> [Mann-Whitney] p-value: {u_pval:.4e} → {'Reject H_0' if u_pval < 0.05 else 'H_0 not rejected'}")
    print("-" * 60)
```
